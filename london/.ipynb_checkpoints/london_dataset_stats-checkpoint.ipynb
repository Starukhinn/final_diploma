```python
columns_to_plot = [
    "car_amenities", "food_amenities", "other",
    "pet_amenity", "shops_amenities",
    "parking_area", "traffic_count", 'сharger_number',
    'apartments', 'aviary',
       'boat', 'booth', 'bunker', 'chapel', 'church', 'club_house',
       'commercial', 'community_centre', 'construction', 'dormitory', 'farm',
       'garage', 'garages', 'gazebo', 'government', 'house', 'hut',
       'industrial', 'kiosk', 'mosque', 'office', 'other_buildings', 'parking',
       'pavilion', 'public', 'railway_arches', 'residential', 'retail', 'roof',
       'school', 'semidetached_house', 'service', 'shed', 'spire',
       'static_caravan', 'synagogue', 'telephone_booth', 'temple', 'terrace',
       'university', 'warehouse', 'aerialway', 'highway', 'port',
       'public_transport', 'railway', 'subway', 'voltage'
]

plt.figure(figsize=(20, 45))

for i, col in enumerate(columns_to_plot, 1):
    plt.subplot(21, 3, i)  
    sns.histplot(df[df[col] != 0][col], bins=30)  
    plt.title(f"Распределение {col}")
    plt.xlabel(col)
    plt.ylabel("Количество")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(20, 45))

for i, col in enumerate(columns_to_plot, 1):
    plt.subplot(21, 3, i)  
    sns.boxplot(x=df[df[col] != 0][col])  
    plt.title(f"Boxplot {col}")
    plt.xlabel(col)

plt.tight_layout()
plt.show()
```

```python
from geopy.distance import geodesic

london_center = (51.5074, -0.1278)

def calculate_distance(lat, lon):
    return geodesic((lat, lon), london_center).km

df["distance_to_center"] = df.apply(lambda row: calculate_distance(row["latitude"], row["longitude"]), axis=1)

df[["latitude", "longitude", "distance_to_center"]]
```

```python
import numpy as np
import pandas as pd

num_classes = 1000

df["distance_class"] = pd.qcut(df["distance_to_center"], num_classes, labels=False, duplicates="drop")


aggregated_data = {}

for col in columns_to_plot:
    filtered_df = df[df[col] > 0]
    
    aggregated = filtered_df.groupby("distance_class")[col].agg(["mean", "median"])
    
    aggregated_data[col] = aggregated

aggregated_df = pd.concat(aggregated_data, axis=1)

aggregated_df.head()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(20, 200))

for i, col in enumerate(columns_to_plot, 1):
    plt.subplot(61, 3, i)

    data = aggregated_df[col]

    sns.scatterplot(x=aggregated_df.index, y=data["mean"], label="Mean", alpha=0.6)
    sns.scatterplot(x=aggregated_df.index, y=data["median"], label="Median", alpha=0.6)

    plt.title(f"Изменение {col} с расстоянием")
    plt.xlabel("Класс расстояния")
    plt.ylabel(col)
    plt.legend()

plt.tight_layout()
plt.show()
```

```python
aggregated_data = {}

for col in columns_to_plot:
    filtered_df = df[df[col] > 0]
    
    aggregated = filtered_df.groupby("distance_class")[col].agg([
        "mean", "median", "std", "var", "min", "max",
        lambda x: x.quantile(0.25), 
        lambda x: x.quantile(0.75),  
        lambda x: x.quantile(0.90), 
        "count", "sum"
    ]).reset_index()
    
    aggregated.columns = [
        "distance_class", "mean", "median", "std", "var", "min", "max",
        "q25", "q75", "q90", "count", "sum"
    ]
    
    aggregated_data[col] = aggregated

aggregated_df = pd.concat(aggregated_data, axis=1)

aggregated_df.head()
```
