```python
import folium
import pandas as pd
import branca.colormap as cm
```

```python
# Создаем карту
m_roads = folium.Map(location=[51.5074, -0.1278], zoom_start=10)

colormap = cm.LinearColormap(
    colors=["green", "yellow", "orange", "red", "darkred"],
    vmin=daily_flow["all_motor_vehicles"].min(),
    vmax=daily_flow["all_motor_vehicles"].max(),
    caption="Traffic Intensity (vehicles/day)")

for _, row in daily_flow.iterrows():
    if pd.notna(row['longitude']) and pd.notna(row['latitude']):
        traffic = row['all_motor_vehicles']
        road_color = colormap(traffic)
        
        folium.CircleMarker(
            location=[row['latitude'], row['longitude']],
            radius=max(traffic / 10000, 2),  # Масштабируем размер точки
            color=road_color,
            fill=True,
            fill_color=road_color,
            fill_opacity=0.6,
            popup=f"{row['road_name']} - {traffic} vehicles/day"
        ).add_to(m_roads)

# Сохраняем карту
m_roads.save("year_traffic.html")
```

```python
import h3
import pandas as pd
import folium
import numpy as np
import matplotlib.pyplot as plt
from scipy.interpolate import griddata
from matplotlib.cm import get_cmap
from matplotlib.colors import to_hex
from matplotlib import colormaps
```

```python
# Загружаем данные
AVERAGE_ANNUAL_DAILY_FLOW_PATH = "/Users/dmitrystarukhin/final_diploma_mipt_sumgf/london/london_data/dft_aadf_region_id_6.csv"
daily_flow = pd.read_csv(AVERAGE_ANNUAL_DAILY_FLOW_PATH).dropna(subset=['latitude', 'longitude'])
```

```python
H3_RESOLUTION = 8  # Более крупные ячейки для лучшей аппроксимации

# Присваиваем каждой точке H3 ячейку
daily_flow['h3_cell'] = daily_flow.apply(lambda row: h3.latlng_to_cell(row['latitude'], row['longitude'], H3_RESOLUTION), axis=1)

# Агрегируем трафик внутри H3 ячеек
traffic_by_h3 = daily_flow.groupby('h3_cell')['all_motor_vehicles'].sum().reset_index()

# Собираем все возможные H3 ячейки в пределах области
unique_h3_cells = set(traffic_by_h3['h3_cell'])
for cell in traffic_by_h3['h3_cell']:
    unique_h3_cells.update(h3.grid_disk(cell, 2))  # Расширяем сетку вокруг известных точек

# Создаем DataFrame со всеми ячейками
all_h3_cells = pd.DataFrame({'h3_cell': list(unique_h3_cells)})

# Заполняем пропущенные ячейки аппроксимацией
known_cells = traffic_by_h3.set_index('h3_cell')['all_motor_vehicles']
all_h3_cells['all_motor_vehicles'] = all_h3_cells['h3_cell'].map(known_cells)

# Аппроксимация пустых значений (интерполяция методом ближайших соседей)
all_h3_cells['all_motor_vehicles'] = all_h3_cells['all_motor_vehicles'].interpolate(method='nearest')
all_h3_cells['all_motor_vehicles'] = all_h3_cells['all_motor_vehicles'].bfill().ffill()

# Нормализация трафика для градиента цветов
all_h3_cells['normalized_traffic'] = (all_h3_cells['all_motor_vehicles'] - all_h3_cells['all_motor_vehicles'].min()) / \
                                      (all_h3_cells['all_motor_vehicles'].max() - all_h3_cells['all_motor_vehicles'].min())
```

```python
all_h3_cells
```

```python
# Выбираем более плавную цветовую карту
cmap = colormaps.get_cmap('plasma')

def get_color(value):
    return to_hex(cmap(value)[:3])  # Конвертируем в HEX
```

```python
# Создаем карту
m_h3 = folium.Map(location=[51.5074, -0.1278], zoom_start=10)

# Добавляем H3 ячейки
for _, row in all_h3_cells.iterrows():
    h3_cell = row['h3_cell']
    hex_boundary = h3.cell_to_boundary(h3_cell)  # Границы ячейки
    hex_boundary = [(lat, lng) for lat, lng in hex_boundary]  # Преобразуем формат

    folium.Polygon(
        locations=hex_boundary,
        color=get_color(row['normalized_traffic']),
        fill=True,
        fill_color=get_color(row['normalized_traffic']),
        fill_opacity=0.6,
        popup=f"Traffic: {row['all_motor_vehicles']} vehicles/day"
    ).add_to(m_h3)

# Отображаем карту
m_h3.save("h3_plasma_year_traffic.html")
```

```python
import h3
import pandas as pd
import folium
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import to_hex
```

```python
# Загружаем данные
AVERAGE_ANNUAL_DAILY_FLOW_PATH = "/Users/dmitrystarukhin/final_diploma_mipt_sumgf/london/london_data/dft_aadf_region_id_6.csv"
daily_flow = pd.read_csv(AVERAGE_ANNUAL_DAILY_FLOW_PATH).dropna(subset=['latitude', 'longitude'])

H3_RESOLUTION = 8  # Детализация сетки H3

# Присваиваем каждой точке H3 ячейку
daily_flow['h3_cell'] = daily_flow.apply(lambda row: h3.latlng_to_cell(row['latitude'], row['longitude'], H3_RESOLUTION), axis=1)


# Агрегируем трафик внутри H3 ячеек
traffic_by_h3 = daily_flow.groupby('h3_cell')['all_motor_vehicles'].sum().reset_index()

# Собираем все возможные H3 ячейки в пределах области
unique_h3_cells = set(traffic_by_h3['h3_cell'])
for cell in traffic_by_h3['h3_cell']:
    unique_h3_cells.update(h3.grid_disk(cell, 2))  # Расширяем сетку вокруг известных точек

    
    
# Создаем DataFrame со всеми ячейками
all_h3_cells = pd.DataFrame({'h3_cell': list(unique_h3_cells)})

# Заполняем пропущенные ячейки аппроксимацией
known_cells = traffic_by_h3.set_index('h3_cell')['all_motor_vehicles']
all_h3_cells['all_motor_vehicles'] = all_h3_cells['h3_cell'].map(known_cells)


# Аппроксимация пустых значений (интерполяция методом ближайших соседей)
all_h3_cells['all_motor_vehicles'] = all_h3_cells['all_motor_vehicles'].interpolate(method='nearest')
all_h3_cells['all_motor_vehicles'] = all_h3_cells['all_motor_vehicles'].bfill().ffill()


# Нормализация трафика для градиента цветов
all_h3_cells['normalized_traffic'] = (all_h3_cells['all_motor_vehicles'] - all_h3_cells['all_motor_vehicles'].min()) / \
                                      (all_h3_cells['all_motor_vehicles'].max() - all_h3_cells['all_motor_vehicles'].min())
```

```python
# Функция цвета (чем больше трафик, тем краснее)
def get_color(value):
    return to_hex((value, 1 - value, 0))  # Градиент от зеленого к красному
```

```python
# Создаем карту
m_h3 = folium.Map(location=[51.5074, -0.1278], zoom_start=10)

# Добавляем H3 ячейки
for _, row in all_h3_cells.iterrows():
    h3_cell = row['h3_cell']
    hex_boundary = h3.cell_to_boundary(h3_cell)  # Границы ячейки
    hex_boundary = [(lat, lng) for lat, lng in hex_boundary]  # Преобразуем формат

    folium.Polygon(
        locations=hex_boundary,
        color=get_color(row['normalized_traffic']),
        fill=True,
        fill_color=get_color(row['normalized_traffic']),
        fill_opacity=0.6,
        popup=f"Traffic: {row['all_motor_vehicles']} vehicles/day"
    ).add_to(m_h3)

# Отображаем карту
m_h3.save("h3_red_green_year_traffic.html")
```

```python
import geopandas as gpd
import folium

# Загружаем данные (замени на свой путь к .gpkg)
file_path = "/Users/dmitrystarukhin/final_diploma_mipt_sumgf/london/Rapid_charging_points.gpkg"
layer_name = "geometry"  # Замени на нужный слой

# Читаем GeoPackage
gdf = gpd.read_file(file_path, layer=None)

# Определяем средний центр для карты
center = [gdf["latitude"].astype(str).str.replace(",", ".").astype(float).mean(), gdf["longtitude"].astype(str).str.replace(",", ".").astype(float).mean()]

# Создаём карту с центром в Лондоне
m = folium.Map(location=center, zoom_start=11)

# Добавляем точки на карту
for _, row in gdf.iterrows():
    folium.Marker(
        location=[row["latitude"], row["longtitude"]],
        popup=row["sitename"],
        tooltip=row["borough"]
    ).add_to(m)

# Показываем карту
m.save('chargers.html')
```
