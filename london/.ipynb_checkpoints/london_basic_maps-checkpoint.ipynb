```python
import requests
import time

# API ключ
api_key = "DSSsfg7CsCUHeFW8qOjlObKgDLGekv80"
base_url = "https://api.tomtom.com/traffic/services/4/flowSegmentData/absolute/10/json"

# Границы Лондона
lat_min, lat_max = 51.286760, 51.691874  # Широта
lon_min, lon_max = -0.510375, 0.334015  # Долгота

# Шаг сетки
step = 0.01  # Чем меньше, тем больше точек (детализация выше)

# Обход всех точек в пределах города
results = []
lat = lat_min
while lat <= lat_max:
    lon = lon_min
    while lon <= lon_max:
        point = f"{lat},{lon}"
        params = {"key": api_key, "point": point}
        response = requests.get(base_url, params=params)
        
        if response.status_code == 200:
            data = response.json()
            results.append(data)
            print(f"Данные для точки {point} успешно получены.")
        else:
            print(f"Ошибка для точки {point}: {response.status_code}, {response.text}")
        
        # Небольшая задержка, чтобы не перегружать API
        time.sleep(0.1)
        
        lon += step
    lat += step

# Сохранение результатов
import json
with open("london_traffic.json", "w") as f:
    json.dump(results, f, indent=4)
```

```python
import pandas as pd
import requests

api_key = "DSSsfg7CsCUHeFW8qOjlObKgDLGekv80"
base_url = "https://api.tomtom.com/traffic/services/4/flowSegmentData/absolute/10/json"

# Пример точки (широта и долгота)
point = "51.5074,-0.1278"

response = requests.get(base_url, params={"point": point, "key": api_key})
if response.status_code == 200:
    data = response.json()
    # Преобразуем данные в DataFrame
    df = pd.DataFrame(data["flowSegmentData"]["coordinates"]["coordinate"])
    df["currentSpeed"] = data["flowSegmentData"]["currentSpeed"]
    df["freeFlowSpeed"] = data["flowSegmentData"]["freeFlowSpeed"]
    print(df.head())
else:
    print(f"Ошибка: {response.status_code}, {response.text}")
```

```python
import requests

# Настройки API
api_key = "DSSsfg7CsCUHeFW8qOjlObKgDLGekv80"
base_url = "https://api.tomtom.com/map/1/staticimage"

# Параметры запроса
params = {
    "key": api_key,
    "zoom": 2,  # Масштаб карты (от 0 до 20)
    "center": "13.5074,-0.1278",  # Координаты центра (широта, долгота)
    "width": 1024,  # Ширина изображения
    "height": 512  # Высота изображения
}

# Выполнение запроса
response = requests.get(base_url, params=params)

# Проверка успешности запроса
if response.status_code == 200:
    # Сохранение изображения в файл
    with open("static_map.png", "wb") as file:
        file.write(response.content)
    print("Карта успешно сохранена как 'static_map.png'")
else:
    print(f"Ошибка: {response.status_code}, {response.text}")
```
