```python
import h3
import folium
import pandas as pd
from shapely.geometry import Polygon

# Координаты центра Сиднея
london_lat, london_lng = -33.840233, 151.01
resolution = H3_RESOLUTION  # Уровень разрешения H3

# Получаем H3 индекс центральной точки
central_hex = h3.latlng_to_cell(london_lat, london_lng, resolution)

# Получаем все шестиугольники в радиусе (k=35 - можно увеличить для покрытия всей зоны)
h3_hexes = h3.grid_disk(central_hex, k=300)

# Список для хранения данных
data = []

# Создаем карту
m = folium.Map(location=[london_lat, london_lng], zoom_start=12)

# Визуализируем шестиугольники и создаем датасет
for hex in h3_hexes:
    hex_boundary = h3.cell_to_boundary(hex)
    # Сохраняем координаты многоугольника в формате списка (широта, долгота)
    for lat, lng in hex_boundary:
        data.append({'latitude': lat, 'longitude': lng, 'h3_index': hex})

# Создаем Pandas DataFrame
df = pd.DataFrame(data)

df
```

```python
df = df.groupby('h3_index', as_index=False).agg({
    'latitude': 'mean',
    'longitude': 'mean',
    **{col: 'first' for col in df.columns if col not in ['latitude', 'longitude', 'h3_index']}
})
```

```python
# Самая южная (минимальная широта)
southernmost_point = df.loc[df['latitude'].idxmin()]

# Самая северная (максимальная широта)
northernmost_point = df.loc[df['latitude'].idxmax()]

# Самая западная (минимальная долгота)
westernmost_point = df.loc[df['longitude'].idxmin()]

# Самая восточная (максимальная долгота)
easternmost_point = df.loc[df['longitude'].idxmax()]

# Самая северо-восточная (максимальная широта, при равных — максимальная долгота)
northeastern_point = df.loc[df.sort_values(by=['latitude', 'longitude'], ascending=[False, False]).index[0]]

# Самая северо-западная (максимальная широта, при равных — минимальная долгота)
northwestern_point = df.loc[df.sort_values(by=['latitude', 'longitude'], ascending=[False, True]).index[0]]

# Самая юго-западная (минимальная широта, при равных — минимальная долгота)
southwestern_point = df.loc[df.sort_values(by=['latitude', 'longitude'], ascending=[True, True]).index[0]]

# Самая юго-восточная (минимальная широта, при равных — максимальная долгота)
southeastern_point = df.loc[df.sort_values(by=['latitude', 'longitude'], ascending=[True, False]).index[0]]

# Вывод результатов
print(f"Самая южная точка: {southernmost_point['latitude']}, {southernmost_point['longitude']}")
print(f"Самая северная точка: {northernmost_point['latitude']}, {northernmost_point['longitude']}")
print(f"Самая западная точка: {westernmost_point['latitude']}, {westernmost_point['longitude']}")
print(f"Самая восточная точка {easternmost_point['latitude']}, {easternmost_point['longitude']}")
print(f"Самая северо-восточная точка: {northeastern_point['latitude']}, {northeastern_point['longitude']}")
print(f"Самая северо-западная точка: {northwestern_point['latitude']}, {northwestern_point['longitude']}")
print(f"Самая юго-западная точка: {southwestern_point['latitude']}, {southwestern_point['longitude']}")
print(f"Самая юго-восточная точка: {southeastern_point['latitude']}, {southeastern_point['longitude']}")
```

```python
import geopandas as gpd

# Добавляем столбцы в df с нулями
categories = ['shops_amenities', 'car_amenities', 'leisure_amenities',
       'needs_amenities', 'food_amenities', 'social_amenities',
       'water_amenities', 'infrastructure_amenities', 'health_amenity',
       'other_amenities', 'home_amenities',
       'children_amenity', 'study_amenity', 'work_amenity', 'pet_amenity',
       'public_transport', 'bike_amenities']

# Функция для определения, к какой H3-ячейке относится объект POI
def get_h3_index(point, resolution=H3_RESOLUTION):
    return h3.latlng_to_cell(point.y, point.x, resolution)

# Добавляем колонку с H3-ячейками в poi_data
poi_data['h3_index'] = poi_data['geometry'].apply(get_h3_index)

# Группируем по H3-ячейкам и категориям, подсчитываем количество объектов
poi_counts = poi_data.groupby(['h3_index', 'category']).size().reset_index(name='count')

# Разворачиваем таблицу: категории становятся столбцами, h3_index остаётся индексом
poi_pivot = poi_counts.pivot_table(index='h3_index', columns='category', values='count', fill_value=0)

# Сбрасываем индекс, чтобы h3_index снова стал столбцом
poi_pivot.reset_index(inplace=True)

# Убедимся, что category не является заголовком столбцов
poi_pivot.columns.name = None  # Убираем имя столбцов, если оно есть
```

```python
# Объединяем df и poi_pivot по h3_index
df = df.merge(poi_pivot, on='h3_index', how='left')

# Заполняем пропущенные значения нулями, если в какой-то клетке H3 нет объектов
df.fillna(0, inplace=True)
```

```python
parking_data = gpd.read_file('/Users/dmitrystarukhin/final_diploma_mipt_sumgf/nsw/data/infrastructure_data/AU-NSW-59c0c26f-20250213-ru-gpkg/data/parking-polygon.gpkg')
parking_data.columns = parking_data.columns.str.lower()

parking_data = parking_data[
    (parking_data.geometry.bounds['miny'] >= lat_min) &
    (parking_data.geometry.bounds['maxy'] <= lat_max) &
    (parking_data.geometry.bounds['minx'] >= lon_min) &
    (parking_data.geometry.bounds['maxx'] <= lon_max)
]
```

```python
import geopandas as gpd
import pandas as pd
import h3
from shapely.geometry import Polygon

# Преобразуем H3 индексы в полигоны
df['geometry'] = df['h3_index'].apply(lambda x: Polygon([(lon, lat) for lat, lon in h3.cell_to_boundary(x)]))

df = gpd.GeoDataFrame(df, geometry='geometry', crs="EPSG:4326")

parking_data = parking_data.explode(ignore_index=True)  # Разделение MultiPolygon на Polygon

# Преобразуем парковочные данные в ту же систему координат
df = df.to_crs("EPSG:3857")
parking_data = parking_data.to_crs(df.crs)

invalid_parking_data = parking_data[~parking_data.is_valid]
invalid_df = df[~df.is_valid]
parking_data['geometry'] = parking_data['geometry'].apply(lambda x: x.buffer(0) if not x.is_valid else x)

# Добавляем колонку с площадью парковок
parking_data['area'] = parking_data['geometry'].area   # Площадь в квадратных метрах

# Находим пересечение парковок с H3 ячейками
parking_intersection = gpd.overlay(parking_data, df, how='intersection')

# Группируем и суммируем площадь парковок по h3_index
parking_sums = parking_intersection.groupby('h3_index')['area'].sum().reset_index()

# Объединяем с df, если в h3_index нет парковки — заполняем 0
df = df.merge(parking_sums, on='h3_index', how='left').fillna({'area': 0})

# Переименуем колонку в parking_area (м²)
df.rename(columns={'area': 'parking_area'}, inplace=True)
df = df.to_crs("EPSG:4326")
```

```python
from shapely.geometry import box

bbox = box(lon_min, lat_min, lon_max, lat_max)
```

```python
power_line = gpd.read_file('/Users/dmitrystarukhin/final_diploma_mipt_sumgf/nsw/data/infrastructure_data/AU-NSW-59c0c26f-20250213-ru-gpkg/data/power-line.gpkg')
power_line.columns = power_line.columns.str.lower()

power_line = power_line[power_line.geometry.type.isin(['LineString', 'MultiLineString'])]

power_line["geometry"] = power_line.geometry.apply(lambda g: g.intersection(bbox))

power_line = power_line[~power_line.geometry.is_empty & power_line.geometry.is_valid]
```

```python
power_line
```

```python
def get_h3_cells(geometry, resolution):
    h3_cells = set()
    if geometry.geom_type == "LineString":  
        for point in geometry.coords:
            h3_cells.add(h3.latlng_to_cell(point[1], point[0], resolution))
    elif geometry.geom_type == "MultiLineString":  
        for line in geometry.geoms:
            for point in line.coords:
                h3_cells.add(h3.latlng_to_cell(point[1], point[0], resolution))
    return list(h3_cells)
```

```python
power_line["h3_index"] = power_line["geometry"].apply(lambda geom: get_h3_cells(geom, H3_RESOLUTION))
power_line_exploded = power_line.explode("h3_index")
power_line_exploded["voltage"] = pd.to_numeric(power_line_exploded["voltage"], errors="coerce")
power_voltage = power_line_exploded.groupby("h3_index")["voltage"].max().reset_index()
df = df.merge(power_voltage, on="h3_index", how="left").fillna(0)
```
