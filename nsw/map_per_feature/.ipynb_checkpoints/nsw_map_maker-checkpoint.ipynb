```python
import geopandas as gpd
import folium
from folium.plugins import MarkerCluster
import h3
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import os
from shapely.geometry import box
from shapely.geometry import Point
```

```python
df = pd.read_csv('/Users/dmitrystarukhin/final_diploma_mipt_sumgf/nsw/data/final_datasets/nsw_final_dataset.csv')
```

```python
categories = ['traffic_count',
       'bike_amenities', 'car_amenities', 'children_amenity', 'food_amenities',
       'health_amenity', 'home_amenities', 'infrastructure_amenities',
       'leisure_amenities', 'needs_amenities', 'other_amenities',
       'pet_amenity', 'public_transport', 'shops_amenities',
       'social_amenities', 'study_amenity', 'water_amenities', 'work_amenity',
     'highway']
```

```python
output_dir = '/Users/dmitrystarukhin/final_diploma_mipt_sumgf/nsw/map_per_feature/maps/amenities'

for category in categories:
    moscow_map = folium.Map(location=[-33.865063, 151.216422], zoom_start=12)
    marker_cluster = MarkerCluster().add_to(moscow_map)
    for idx, row in df.iterrows():
        if row[category]:
            folium.Marker(
                    location=[row.latitude, row.longitude],
                    popup=row.get('amenity', f'{category}'),
                    tooltip=row[category],
                    icon=folium.Icon(color='blue', icon='info-sign')
                ).add_to(marker_cluster)
    output_path = os.path.join(output_dir, f"nsw_{category}_map.html")
    moscow_map.save(output_path)
    print(f"Карта {category} сохранена")
```

```python
# Примерные границы Сиднея по широте и долготе
lat_min, lat_max = -34.2183, -33.495512  # Примерные границы по широте
lon_min, lon_max = 150.5200, 151.3433  # Примерные границы по долготе
```

```python
parking_data = gpd.read_file('/Users/dmitrystarukhin/final_diploma_mipt_sumgf/nsw/data/infrastructure_data/AU-NSW-59c0c26f-20250213-ru-gpkg/data/parking-polygon.gpkg')
parking_data.columns = parking_data.columns.str.lower()


parking_data = parking_data[
    (parking_data.geometry.bounds['miny'] >= lat_min) &
    (parking_data.geometry.bounds['maxy'] <= lat_max) &
    (parking_data.geometry.bounds['minx'] >= lon_min) &
    (parking_data.geometry.bounds['maxx'] <= lon_max)
]
```

```python
parking_map = folium.Map(location=[-33.865063, 151.216422], zoom_start=12, tiles='cartodb positron', max_bounds=True)

for _, row in parking_data.iterrows():
    folium.GeoJson(row['geometry'], style_function=lambda x: {'color': 'blue', 'opacity': 0.5}).add_to(parking_map)

output_dir = '/Users/dmitrystarukhin/final_diploma_mipt_sumgf/nsw/map_per_feature/maps'
output_path = os.path.join(output_dir, f"nsw_parkings.html")
parking_map.save(output_path)
```

```python
chargers_map = folium.Map(location=[-33.865063, 151.216422], zoom_start=12)
marker_cluster = MarkerCluster().add_to(chargers_map)

for _, row in df.iterrows():
    if row['has_charging_station']:
        folium.Marker(
            location=[row['latitude'], row['longitude']],
            popup=f"Stations: {row['number_stations']}",
            icon=folium.Icon(color='blue', icon='bolt', prefix='fa'),
            tooltip="Charging Station"
        ).add_to(chargers_map)

output_path = os.path.join(output_dir, f"nsw_chargers_map.html")
chargers_map.save(output_path)
```

```python
from shapely.geometry import box

# Примерные границы Сиднея по широте и долготе
lat_min, lat_max = -34.2183, -33.495512  # Примерные границы по широте
lon_min, lon_max = 150.5200, 151.3433  # Примерные границы по долготе

bbox = box(lon_min, lat_min, lon_max, lat_max)
```

```python
power_line = gpd.read_file('/Users/dmitrystarukhin/final_diploma_mipt_sumgf/nsw/data/infrastructure_data/AU-NSW-59c0c26f-20250213-ru-gpkg/data/power-line.gpkg')
power_line.columns = power_line.columns.str.lower()

power_line = power_line[power_line.geometry.type.isin(['LineString', 'MultiLineString'])]

power_line["geometry"] = power_line.geometry.apply(lambda g: g.intersection(bbox))

power_line = power_line[~power_line.geometry.is_empty & power_line.geometry.is_valid]
```

```python
output_dir = '/Users/dmitrystarukhin/final_diploma_mipt_sumgf/nsw/map_per_feature/maps'
nsw_map = folium.Map(location=[-33.865063, 151.216422], zoom_start=12)
folium.GeoJson(power_line, name="Power Lines").add_to(nsw_map)
output_path = os.path.join(output_dir, f"nsw_power_lines_map.html")
nsw_map.save(output_path)
```

```python
# Создаем базовую карту с центром в Сиднее
combined_map = folium.Map(location=[-33.865063, 151.216422], zoom_start=12)

# 1. Добавляем зарядные станции (кластеризованные маркеры)
charger_cluster = MarkerCluster(name="Charging Stations").add_to(combined_map)

for _, row in df.iterrows():
    if row['has_charging_station']:
        folium.Marker(
            location=[row['latitude'], row['longitude']],
            popup=f"Stations: {row['number_stations']}",
            icon=folium.Icon(color='blue', icon='bolt', prefix='fa'),
            tooltip="Charging Station"
        ).add_to(charger_cluster)

# 2. Добавляем линии электропередач
power_line = gpd.read_file('/Users/dmitrystarukhin/final_diploma_mipt_sumgf/nsw/data/infrastructure_data/AU-NSW-59c0c26f-20250213-ru-gpkg/data/power-line.gpkg')
power_line.columns = power_line.columns.str.lower()

power_line = gpd.read_file('/Users/dmitrystarukhin/final_diploma_mipt_sumgf/nsw/data/infrastructure_data/AU-NSW-59c0c26f-20250213-ru-gpkg/data/power-line.gpkg')
power_line.columns = power_line.columns.str.lower()

power_line = power_line[power_line.geometry.type.isin(['LineString', 'MultiLineString'])]

power_line["geometry"] = power_line.geometry.apply(lambda g: g.intersection(bbox))

power_line = power_line[~power_line.geometry.is_empty & power_line.geometry.is_valid]

# Стиль для линий электропередач
def style_function(feature):
    return {
        'color': 'red',
        'weight': 3,
        'opacity': 0.7
    }

# Добавляем линии на карту
folium.GeoJson(
    power_line,
    name="Power Lines",
    style_function=style_function,
    tooltip=folium.GeoJsonTooltip(fields=['voltage'], aliases=['Voltage:'])
).add_to(combined_map)

# 3. Добавляем управление слоями
folium.LayerControl().add_to(combined_map)

# Сохраняем карту
output_path = os.path.join(output_dir, "nsw_combined_chargers_power_lines.html")
combined_map.save(output_path)

print(f"Комбинированная карта сохранена: {output_path}")
```
