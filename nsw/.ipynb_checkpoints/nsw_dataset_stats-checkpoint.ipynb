```python
def process_charger_rating(value):
    if value == "nothing":
        return 0  
    value = value.replace(" ", "")  
    values = []
    
    for v in value.split(","):  
        if v == "kW":  
            values.append(1) 
        elif "kW" in v:  
            try:
                num = float(v.split("kW")[0])  
                values.append(num)
            except ValueError:
                continue  

    if not values:
        return 0 

    if "-" in value:  
        min_val, max_val = float(value.split("-")[0]), float(value.split("-")[-1].split("kW")[0])
        return (min_val + max_val) / 2  

    return np.mean(values) 
```

```python
df["charger_rating_float"] = df["charger_rating"].apply(process_charger_rating)
```

```python
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
sns.histplot(df[df["charger_rating_float"] != 0]["charger_rating_float"], bins=30)
plt.title("Распределение энергии зарядки гексагоне")
```

```python
from geopy.distance import geodesic

sydney_center = (-33.8688, 151.2093)

def calculate_distance(lat, lon):
    return geodesic((lat, lon), sydney_center).km

df["distance_to_center"] = df.apply(lambda row: calculate_distance(row["latitude"], row["longitude"]), axis=1)
```

```python
df[["latitude", "longitude", "distance_to_center"]]
```

```python
import numpy as np
import pandas as pd

num_classes = 1000


columns_to_plot = [
    'beauty_amenities', 'bicycle_amenities', 'bike_amenities','car_amenities',
    'children_amenity', 'food_amenities', 'health_amenity', 'home_amenities',
    'infrastructure_amenities', 'leisure_amenities', 'needs_amenities', 'other', 'pet_amenity', 'shops_amenities', 
    'social_amenities', 'social_transport_amenities', 'storage_amenities', 'study_amenity', 'water_amenities', 
    'work_amenity', 'parking_area', 'traffic_count', 'number_stations', 'number_plugs', 'charger_rating_float',
    'apartments', 'brewery',
       'bunker', 'cabin', 'carport', 'church', 'commercial', 'construction',
       'detached', 'dormitory', 'garage', 'government', 'hotel', 'house',
       'hut', 'industrial', 'kindergarten', 'kiosk', 'mosque', 'office',
       'other_buildings', 'pub', 'public', 'residential', 'roof', 'school',
       'semidetached_house', 'service', 'shed', 'temple', 'tent', 'terrace',
       'toilets', 'townhouse', 'warehouse', 'window', 'aerialway', 'highway',
       'port', 'public_transport', 'railway', 'subway', 'voltage'
]


df["distance_class"] = pd.qcut(df["distance_to_center"], num_classes, labels=False, duplicates="drop")

aggregated_data = {}

for col in columns_to_plot:
    filtered_df = df[df[col] > 0]
    
    aggregated = filtered_df.groupby("distance_class")[col].agg(["mean", "median"])
    
    aggregated_data[col] = aggregated

aggregated_df = pd.concat(aggregated_data, axis=1)

aggregated_df.head()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(20, 180))


rows, cols = 24, 3  

for i, col in enumerate(columns_to_plot, 1):
    plt.subplot(rows, cols, i)

    data = aggregated_df[col]

    sns.scatterplot(x=aggregated_df.index, y=data["mean"], label="Mean", alpha=0.6)
    sns.scatterplot(x=aggregated_df.index, y=data["median"], label="Median", alpha=0.6)

    plt.title(f"Изменение {col} с расстоянием")
    plt.xlabel("Класс расстояния")
    plt.ylabel(col)
    plt.legend()

plt.tight_layout()
plt.show()
```

```python
aggregated_data = {}

for col in columns_to_plot:
    filtered_df = df[df[col] > 0]
    
    aggregated = filtered_df.groupby("distance_class")[col].agg([
        "mean", "median", "std", "var", "min", "max",
        lambda x: x.quantile(0.25), 
        lambda x: x.quantile(0.75),  
        lambda x: x.quantile(0.90), 
        "count", "sum"
    ]).reset_index()
    
    aggregated.columns = [
        "distance_class", "mean", "median", "std", "var", "min", "max",
        "q25", "q75", "q90", "count", "sum"
    ]
    
    aggregated_data[col] = aggregated

aggregated_df = pd.concat(aggregated_data, axis=1)

aggregated_df.head()
```
