```python
import folium

# Создаем базовую карту, центрируем её на среднем значении координат
# (например, для начала используем средние значения широты и долготы)
map_center = [traffic_df['Latitude'].mean(), traffic_df['Longitude'].mean()]
traffic_map = folium.Map(location=map_center, zoom_start=10)

# Добавляем метки на карту для каждого сегмента
for idx, row in traffic_df.iterrows():
    lat = row['Latitude']
    lon = row['Longitude']
    road_section_id = row['Road_Section_ID']
    aadt = row['AADT']
    carriageway_code = row['Carriageway_Code']
    
    # Создаем popup для каждой метки с дополнительной информацией
    popup_content = f"Road Section ID: {road_section_id}<br>AADT: {aadt}<br>Carriageway Code: {carriageway_code}"
    
    # Добавляем метку на карту
    folium.Marker(
        location=[lat, lon],
        popup=popup_content,
        icon=folium.Icon(color='blue', icon='info-sign')
    ).add_to(traffic_map)
```

```python
import pandas as pd
import h3
import folium
import numpy as np
from folium import plugins
from matplotlib.colors import to_hex


traffic_df = pd.read_csv("/Users/dmitrystarukhin/Downloads/road-location-and-traffic-data_20230718.csv")
traffic_df = traffic_df.dropna(subset=['AADT', 'Latitude', 'Longitude'])
# traffic_df = traffic_df.iloc[:2000]
# traffic_df = traffic_df.sample(n=1000000, random_state=42)
# traffic_df = traffic_df.copy()


# Преобразуем координаты в ячейки H3
H3_RESOLUTION = 8  # Детализация сетки H3

# Присваиваем каждой точке H3 ячейку
traffic_df['h3_cell'] = traffic_df.apply(lambda row: h3.latlng_to_cell(row['Latitude'], row['Longitude'], H3_RESOLUTION), axis=1)

# Агрегируем трафик внутри H3 ячеек
traffic_by_h3 = traffic_df.groupby('h3_cell')['AADT'].sum().reset_index()

# Собираем все возможные H3 ячейки в пределах области
unique_h3_cells = set(traffic_by_h3['h3_cell'])
for cell in traffic_by_h3['h3_cell']:
    unique_h3_cells.update(h3.grid_disk(cell, 2))  # Расширяем сетку вокруг известных точек

# Создаем DataFrame со всеми ячейками
all_h3_cells = pd.DataFrame({'h3_cell': list(unique_h3_cells)})

# Заполняем пропущенные ячейки аппроксимацией
known_cells = traffic_by_h3.set_index('h3_cell')['AADT']
all_h3_cells['AADT'] = all_h3_cells['h3_cell'].map(known_cells)

# Аппроксимация пустых значений (интерполяция методом ближайших соседей)
all_h3_cells['AADT'] = all_h3_cells['AADT'].interpolate(method='nearest')
all_h3_cells['AADT'] = all_h3_cells['AADT'].bfill().ffill()

# Нормализация трафика для градиента цветов
aadt_min = all_h3_cells['AADT'].min()
aadt_max = all_h3_cells['AADT'].max()

if aadt_min == aadt_max:
    all_h3_cells['normalized_traffic'] = 0.01 
else:
    all_h3_cells['normalized_traffic'] = (all_h3_cells['AADT'] - 100) / (70000 - 100)
```

```python
def get_color(value):
    value = max(0, min(1, value))  # Ограничиваем в пределах [0,1]
    cmap = cm.get_cmap('RdYlGn_r')  # Обратная цветовая карта (красный → жёлтый → зелёный)
    return mcolors.to_hex(cmap(value))
```

```python
from matplotlib import colormaps

cmap = colormaps.get_cmap('plasma')

def get_color(value):
    return to_hex(cmap(value)[:3])  # Конвертируем в HEX
```

```python
# Создаем карту с центром на среднем значении координат
map_center = [traffic_df['Latitude'].mean(), traffic_df['Longitude'].mean()]
traffic_h3 = folium.Map(location=map_center, zoom_start=10)

# Добавляем H3 ячейки на карту
for _, row in all_h3_cells.iterrows():
    h3_cell = row['h3_cell']
    hex_boundary = h3.cell_to_boundary(h3_cell)  # Границы ячейки
    hex_boundary = [(lat, lng) for lat, lng in hex_boundary]  # Преобразуем формат

    # Добавляем ячейку на карту с цветом, основанным на трафике
    folium.Polygon(
        locations=hex_boundary,
        color=get_color(row['normalized_traffic']),
        fill=True,
        fill_color=get_color(row['normalized_traffic']),
        fill_opacity=0.8,
        popup=f"AADT: {row['AADT']}"
    ).add_to(traffic_h3)
```
