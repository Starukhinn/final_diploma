```python
# Цветовая схема в зависимости от количества удобств
max_count = 10

def get_colors(counts, max_count):
    counts = np.array(counts)  # Преобразуем в NumPy массив

    # Нормализуем значения
    norm_counts = counts / max_count if max_count > 0 else 0

    # Получаем цветовую карту
    colormap = plt.get_cmap("RdYlGn_r")  
    rgba = colormap(norm_counts)  # Применяем к каждому элементу

    # Преобразуем в строковый формат RGBA
    colors = [f'rgba({int(r[0] * 255)}, {int(r[1] * 255)}, {int(r[2] * 255)}, {r[3]})' for r in rgba]
    
    return colors
```

```python
h3_counts['color'] = get_colors(h3_counts['count'], max_count)

# Добавление ячеек H3 на карту
for _, row in h3_counts.iterrows():
    # Получаем границы ячейки
    boundary = h3.cell_to_boundary(row['h3_index'])
    
    # Преобразуем в формат, который понимает folium
    polygon = [[(lat, lng) for lat, lng in boundary]]

    folium.Polygon(
        locations=polygon,
        color=row['color'],  # Используем предрассчитанный цвет
        fill=True,
        fill_opacity=0.6,
        popup=f'Count: {row["count"]}'
    ).add_to(m)

# Добавление парковок
for _, row in parking_data.iterrows():
    folium.GeoJson(
        row['geometry'],
        style_function=lambda x: {'color': 'blue', 'opacity': 0.5}
    ).add_to(m)
```

```python
import geopandas as gpd
import folium
import h3
import numpy as np
import matplotlib.pyplot as plt
from shapely.geometry import Polygon

# Загрузка данных
parking_data = gpd.read_file('/Users/dmitrystarukhin/final_diploma_mipt_sumgf/nsw/AU-NSW-59c0c26f-20250213-ru-gpkg/data/parking-polygon.gpkg')
poi_data = gpd.read_file('/Users/dmitrystarukhin/final_diploma_mipt_sumgf/nsw/AU-NSW-59c0c26f-20250213-ru-gpkg/data/poi-point.gpkg')

poi_data.columns = poi_data.columns.str.lower()
parking_data.columns = parking_data.columns.str.lower()

# Объединение всех категорий
all_amenities = food_amenities + shops_amenities + car_amenities + bike_amenities + leisure_amenities + \
                 social_transport_amenities + leisure_amenities + social_amenities + needs_amenities + \
                 bicycle_amenities + water_amenities + home_amenities + study_amenity + children_amenity + \
                 work_amenity + beauty_amenities + storage_amenities + health_amenity + infrastructure_amenities + pet_amenity

# Фильтрация POI по категориям
filtered_poi_data = poi_data[poi_data['amenity'].isin(all_amenities)].copy()

# Создание H3 индексов
filtered_poi_data['h3_index'] = filtered_poi_data['geometry'].apply(lambda x: h3.latlng_to_cell(x.y, x.x, 6))

# Подсчет количества удобств в каждом H3 индексе
h3_counts = filtered_poi_data['h3_index'].value_counts().reset_index()
h3_counts.columns = ['h3_index', 'count']
h3_counts = h3_counts.set_index('h3_index')

# Заполняем пропущенные H3 индексы
all_h3_cells = set()
for h3_index in h3_counts.index:
    all_h3_cells.add(h3_index)
    all_h3_cells.update(h3.grid_disk(h3_index, 1))  # Добавляем соседей

# Добавляем отсутствующие ячейки с нулевыми значениями
for h3_index in all_h3_cells:
    if h3_index not in h3_counts.index:
        h3_counts.loc[h3_index] = 0

# Функция для интерполяции пустых ячеек
for h3_index in h3_counts.index:
    if h3_counts.loc[h3_index, 'count'] == 0:
        neighbors = h3.grid_disk(h3_index, 1)
        neighbor_values = [h3_counts.loc[n, 'count'] for n in neighbors if n in h3_counts.index and h3_counts.loc[n, 'count'] > 0]
        if neighbor_values:
            h3_counts.loc[h3_index, 'count'] = int(np.mean(neighbor_values))

# Генерация цветовой гаммы
max_count = h3_counts['count'].max()
def get_colors(counts, max_count):
    norm_counts = np.minimum(np.maximum(np.array(counts) - 5, 0) / 40, 1)
    colormap = plt.get_cmap("RdYlGn_r")
    rgba = colormap(norm_counts)
    return [f'rgba({int(r[0] * 255)}, {int(r[1] * 255)}, {int(r[2] * 255)}, {r[3]})' for r in rgba]

h3_counts['color'] = get_colors(h3_counts['count'], max_count)

# Отрисовка карты
m = folium.Map(location=[-31.840233, 145.612793], zoom_start=12, tiles='cartodb positron', max_bounds=True)

# Добавление H3 ячеек
for h3_index, row in h3_counts.iterrows():
    boundary = h3.cell_to_boundary(h3_index)
    polygon = [[(lat, lng) for lat, lng in boundary]]
    folium.Polygon(locations=polygon, color=row['color'], fill=True, fill_opacity=0.6, popup=f'Count: {row["count"]}').add_to(m)

# Добавление парковок
for _, row in parking_data.iterrows():
    folium.GeoJson(row['geometry'], style_function=lambda x: {'color': 'blue', 'opacity': 0.5}).add_to(m)

m.save('nsw_h3_filled.html')
```
