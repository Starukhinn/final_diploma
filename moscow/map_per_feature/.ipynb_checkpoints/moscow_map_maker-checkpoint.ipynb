```python
import geopandas as gpd
import folium
from folium.plugins import MarkerCluster
import h3
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import os
from shapely.geometry import box
from shapely.geometry import Point
```

```python
df = pd.read_csv('/Users/dmitrystarukhin/final_diploma_mipt_sumgf/moscow/moscow_data/final_datasets/dataset_moscow.csv')
```

```python
categories = [
       'bike_amenities', 'car_amenities', 'children_amenity', 'food_amenities',
       'health_amenity', 'home_amenities', 'infrastructure_amenities',
       'leisure_amenities', 'needs_amenities', 'other_amenities',
       'pet_amenity', 'public_transport', 'shops_amenities',
       'social_amenities', 'study_amenity', 'water_amenities', 'work_amenity',
     'highway']
```

```python
output_dir = '/Users/dmitrystarukhin/final_diploma_mipt_sumgf/moscow/map_per_feature/maps/amenities'

for category in categories:
    moscow_map = folium.Map(location=[55.7522, 37.6156], zoom_start=12)
    marker_cluster = MarkerCluster().add_to(moscow_map)
    for idx, row in df.iterrows():
        if row[category]:
            folium.Marker(
                    location=[row.latitude, row.longitude],
                    popup=row.get('amenity', f'{category}'),
                    tooltip=row[category],
                    icon=folium.Icon(color='blue', icon='info-sign')
                ).add_to(marker_cluster)
    output_path = os.path.join(output_dir, f"moscow_{category}_map.html")
    moscow_map.save(output_path)
    print(f"Карта {category} сохранена")
```

```python
# Примерные границы Сиднея по широте и долготе
lat_min, lat_max = -34.2183, -33.495512  # Примерные границы по широте
lon_min, lon_max = 150.5200, 151.3433  # Примерные границы по долготе
```

```python
parking_data = gpd.read_file('/Users/dmitrystarukhin/final_diploma_mipt_sumgf/nsw/data/infrastructure_data/AU-NSW-59c0c26f-20250213-ru-gpkg/data/parking-polygon.gpkg')
parking_data.columns = parking_data.columns.str.lower()

parking_data = parking_data[
    (parking_data.geometry.bounds['miny'] >= lat_min) &
    (parking_data.geometry.bounds['maxy'] <= lat_max) &
    (parking_data.geometry.bounds['minx'] >= lon_min) &
    (parking_data.geometry.bounds['maxx'] <= lon_max)
```

```python
from shapely.geometry import box
from shapely.geometry import Point

# Примерные границы Сиднея по широте и долготе
lat_min, lat_max = -34.2183, -33.495512  # Примерные границы по широте
lon_min, lon_max = 150.5200, 151.3433  # Примерные границы по долготе

bbox = box(lon_min, lat_min, lon_max, lat_max)
def line_in_bbox(geom, bbox):
    if geom.is_empty:
        return False
    if geom.geom_type == 'LineString':
        return any(bbox.contains(Point(point)) for point in geom.coords)
    elif geom.geom_type == 'MultiLineString':
        return any(bbox.contains(Point(point)) for line in geom.geoms for point in line.coords)
    return False
```

```python
parking_map = folium.Map(location=[55.7522, 37.6156], zoom_start=12, tiles='cartodb positron', max_bounds=True)

for _, row in parking_data.iterrows():
    folium.GeoJson(row['geometry'], style_function=lambda x: {'color': 'blue', 'opacity': 0.5}).add_to(parking_map)

output_dir = '/Users/dmitrystarukhin/final_diploma_mipt_sumgf/moscow/map_per_feature/maps'
output_path = os.path.join(output_dir, f"moscow_parkings.html")
parking_map.save(output_path)
```

```python
chargers_map = folium.Map(location=[55.7522, 37.6156], zoom_start=12)
marker_cluster = MarkerCluster().add_to(chargers_map)

for _, row in df.iterrows():
    if row['сharger_number']:
        folium.Marker(
            location=[row['latitude'], row['longitude']],
            popup=f"number: {row['сharger_number']}",
            icon=folium.Icon(color='blue')
        ).add_to(chargers_map)

output_path = os.path.join(output_dir, f"moscow_chargers_map.html")
chargers_map.save(output_path)
```

```python
power_line = gpd.read_file('/Users/dmitrystarukhin/final_diploma_mipt_sumgf/nsw/data/infrastructure_data/AU-NSW-59c0c26f-20250213-ru-gpkg/data/power-line.gpkg')
power_line.columns = power_line.columns.str.lower()

power_line = power_line[power_line.geometry.apply(lambda g: line_in_bbox(g, bbox))]
```

```python
output_dir = '/Users/dmitrystarukhin/final_diploma_mipt_sumgf/moscow/map_per_feature/maps'
moscow_map = folium.Map(location=[55.7522, 37.6156], zoom_start=12)
folium.GeoJson(power_line, name="Power Lines").add_to(moscow_map)
output_path = os.path.join(output_dir, f"moscow_power_lines_map.html")
moscow_map.save(output_path)
```
